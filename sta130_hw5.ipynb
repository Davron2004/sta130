# Question 2

The sentence in question is this:

"It is important to note that outcomes of tests refer to the population parameter, rather than the sample statistic! As such, the result that we get is for the population."

### First, I want to make a clear destinction between "population parameters" and "sample statistics".
#### Population Parameter
A *Population Parameter* is the true value of a certain statistic for *all* of it in the *world*. For example, average height of all the male humans on planet Earth in a *Population Parameter*.
#### Sample Statistic
When we deal with statistics, we don't have *all* of the data entries, instead we have a *sample*, which is just a small subset of what's in the *world*. So a *Sample Statistic* is a statistic for *the subset that we have*. For example, if we had the data about the height values of 1000 male humans, the mean value of their heights would be a *sample statistic*.

### Secondly, let's describe the meaning of the four introduced terms in a way that's understandable for "non-statistics" people
- xᵢ's represent the enumerated set of input data, *sample*. They're relevant to *sample statistics*.
- x̄ is the mean value of xᵢ's. This is a *sample statistic*.
- μ is the *actual* mean value from *all of the population, the world*. It is a *population parameter*. It is unknown; the whole point of statistics is guessing what it (and other population parameters) could be.
- μ₀ is the *hypothesized* value of a *population parameter* μ. It is a *population parameter*.

### Now that we're equipped with the explanations of the technical terms, we can describe the sentence on the top with them.

The statement means that while we use the *sample data* (xᵢ's and x̄) to perform the test, the result tells us whether the *population parameter mean* (μ) is likely to match the *hypothesized value* (μ₀), making it about the population rather than the sample.

"Imagining a world where the null hypothesis is true" is basically saying "what if the actual population parameters are the ones that are predicted and our sample is off". This is useful to see how much a certain datum "disagrees" with what was predicted beforehand, which we call a *p-value*

# Question 4

The smaller the p-value is, the "further away" our data is from what was guessed in the beginning to be average, which means that what we guessed in the beginning is likely to be incorrect. For example, let's say that I have 1000 entries of male height, and let's say that the hypothesized value for the male average height was 100cm. Now I calculate the average male height in my data sample and wow, I get 200cm. Now that I look at it, μ₀ = 100cm looks ridiculously incorrect compared to x̄ = 200cm I got. And of course 200cm has a very small p-value. This is how p-value can make the null hypothesis look "ridiculous"
